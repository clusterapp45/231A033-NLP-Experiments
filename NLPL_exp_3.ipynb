```python
text = '''Hey everyone!!! 👋

I'm REALLY excited 😊 because I've just completed my first Machine Learning project on 23rd July, 2026. The project analyzed over 12,500 customer reviews collected from https://www.example.com and another dataset downloaded from www.kaggle.com.

Unfortunately, some reviews weren't useful—they contained only words like "the", "is", "and", "of", "to", etc. Others had strange symbols such as @#$%^&*, repeated punctuation!!!!!!, HTML tags like <br>, and emojis 😂😂😂.

You can contact me at vijay.shanker@eng.rizvi.edu.in or support123@company.co.in if you'd like the dataset. Follow us on Twitter @AI_ResearchLab and use hashtags like #MachineLearning #NLP #DataScience.

The model's accuracy increased from 82.5% to 91.8%, while the training time decreased by nearly 35%. It processed 1,25,000 records in less than 2.5 hours and consumed around 3.7 GB of RAM.

Students were studying, analyzing, processing, predicting, running, writing, and comparing different algorithms. The researchers discovered better techniques after performing extensive experiments and evaluating multiple models.

Don't worry if the results aren't perfect—you're learning! Let's improve them together. I'll upload the final report by 30/07/2026.

Thanks!!!'''

print("Shaikh Abdullah", "UIN: 231A033\n")
print(text)
```

```python
print("Shaikh Abdullah", "UIN: 231A033\n")

text_lower = text.lower()

print(text_lower)
```

```python
import re

print("Shaikh Abdullah", "UIN: 231A033\n")

pattern = r'https?://\S+|www\.\S+|\S+@\S+'

text_url = re.sub(pattern, " ", text_lower)

print("Text after removing URLs and email addresses:")
print(text_url)
```

```python
print("Shaikh Abdullah", "UIN: 231A033\n")

pattern = r'\b\d[\d,.]*%?\b'

text_percentage = re.sub(pattern, " ", text_url)

print("Text after removing numbers:")
print(text_percentage)
```

```python
import emoji

print("Shaikh Abdullah", "UIN: 231A033\n")

text_emoji = emoji.replace_emoji(text_percentage, replace="")

print("Text after removing emojis:")
print(text_emoji)
```

```python
print("Shaikh Abdullah", "UIN: 231A033\n")

text_html = re.sub(r'<[^>]+>', ' ', text_emoji)

print("Text after removing HTML tags:")
print(text_html)
```

```python
import nltk

nltk.download('punkt')
nltk.download('punkt_tab')

from nltk import word_tokenize

print("Shaikh Abdullah", "UIN: 231A033\n")

tokenized = word_tokenize(text_html)

print("Tokenized words:")
print(tokenized)
```

```python
import string

print("Shaikh Abdullah", "UIN: 231A033\n")

punctuation = list(string.punctuation)

print("Punctuation symbols:")
print(punctuation)
```

```python
print("Shaikh Abdullah", "UIN: 231A033\n")

print("With punctuation:")
print(tokenized)

text_punct = []

for word in tokenized:
    if word not in punctuation:
        text_punct.append(word)

print("\nWithout punctuation:")
print(text_punct)
```

```python
import nltk

nltk.download("stopwords")

from nltk.corpus import stopwords

print("Shaikh Abdullah", "UIN: 231A033\n")

stop_words = stopwords.words("english")

print("English stop words:")
print(stop_words)
```

```python
print("Shaikh Abdullah", "UIN: 231A033\n")

text_clean2 = []

for word in text_punct:
    if word.lower() not in stop_words:
        text_clean2.append(word.lower())

print("Text after stop-word removal:")
print(text_clean2)
```

```python
from nltk.stem import PorterStemmer

print("Shaikh Abdullah", "UIN: 231A033\n")

stemmer = PorterStemmer()

stemmed_words = [stemmer.stem(word) for word in text_clean2]

print("Stemmed words:")
print(stemmed_words)
```

```python
import nltk

nltk.download('wordnet')
nltk.download('omw-1.4')

from nltk.stem import WordNetLemmatizer

print("Shaikh Abdullah", "UIN: 231A033\n")

lemmatizer = WordNetLemmatizer()

lemmatized_words = [
    lemmatizer.lemmatize(word) for word in text_clean2
]

print("Lemmatized words:")
print(lemmatized_words)
```

```python
print("Shaikh Abdullah", "UIN: 231A033\n")

vocab_before_stop_words = len(set(text_punct))
vocab_after_stop_words = len(set(text_clean2))
vocab_after_lemmatization = len(set(lemmatized_words))

print("Vocabulary size before stop-word removal:",
      vocab_before_stop_words)

print("Vocabulary size after stop-word removal:",
      vocab_after_stop_words)

print("Vocabulary size after lemmatization:",
      vocab_after_lemmatization)

print("-" * 30)
```

```python
print("Shaikh Abdullah", "UIN: 231A033\n")

stemmed_words = [stemmer.stem(word) for word in text_clean2]

print("Stemmed words:")
print(stemmed_words)

print("\nLemmatized words:")
print(lemmatized_words)
```

## Assessment of Preprocessing Impact

### 1. Reduction in Token Count
Stop-word removal reduces the total number of tokens by eliminating common words that may contribute little to the selected NLP task. Lemmatization converts words into their base forms.

### 2. Reduction in Vocabulary Size
Stop-word removal and lemmatization can reduce vocabulary size, which may help lower dimensionality in text-processing tasks.

### 3. Stemming vs. Lemmatization

**Stemming:** Porter Stemmer removes word endings using simple rules. It is generally faster but may produce words that are not valid dictionary words.

**Lemmatization:** WordNet Lemmatizer converts words into dictionary forms. It can produce more meaningful results, especially when used with appropriate part-of-speech information.

### Conclusion
Text preprocessing improves the consistency of text data by removing unnecessary elements and reducing noise. Tokenization, punctuation removal, stop-word removal, stemming, and lemmatization prepare text for further Natural Language Processing tasks. The choice between stemming and lemmatization depends on the requirements of the application.
