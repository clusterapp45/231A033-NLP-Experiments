```python
student = "Shaikh Abdullah | UIN: 231A033"
print(student)
import nltk
from nltk.corpus import wordnet as wn

nltk.download('wordnet')
nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
print(student)
def lesk(word, sentence):
    """Implementation of the Lesk algorithm."""

    sentence_words = set(nltk.word_tokenize(sentence.lower()))

    best_sense = None
    max_overlap = 0

    for sense in wn.synsets(word):
        definition = set(
            nltk.word_tokenize(sense.definition().lower())
        )

        examples = set(
            nltk.word_tokenize(
                " ".join(sense.examples()).lower()
            )
        )

        overlap = len(
            sentence_words & (definition | examples)
        )

        if overlap > max_overlap:
            max_overlap = overlap
            best_sense = sense

    return best_sense
```

```python
print(student)
word = "bank"
sentence = "I deposited money in the bank."

sense = lesk(word, sentence)

print("Word:", word)

if sense:
    print("Meaning:", sense.definition())
    print("Example:", sense.examples())
else:
    print("No sense found.")
```

```python
print(student)
from nltk import word_tokenize

d = "i am going to bank"
s = "i work in a bank"
example = "i like to go to bank"

overlap = (
    len(set(word_tokenize(d)).intersection(word_tokenize(example)))
    + len(set(word_tokenize(s)).intersection(word_tokenize(example)))
)

print(overlap)
```

```python
print(student)
sentence = "I went to the river bank to swim."
ambiguous_word = "swim"

sense = lesk(ambiguous_word, sentence)

if sense:
    print(
        f"The best sense for '{ambiguous_word}' "
        f"in the given context is: "
        f"{sense.name()} - {sense.definition()}"
    )
else:
    print("No sense found.")
```

```python
print(student)
dataset = [
    ("I went to the bank to deposit some money.", "bank.n.01"),
    ("He has a bank of knowledge.", "bank.n.02")
]

correct_predictions = 0

for sentence, correct_sense in dataset:
    predicted_sense = lesk("bank", sentence)

    if predicted_sense and predicted_sense.name() == correct_sense:
        correct_predictions += 1

accuracy = correct_predictions / len(dataset)

print(f"Accuracy: {accuracy:.2f}")
```
