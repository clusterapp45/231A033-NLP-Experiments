```python
all_positive_twitts = twitter_samples.strings('positive_tweets.json')
all_negative_twitts = twitter_samples.strings('negative_tweets.json')
print("Shaikh Abdullah - 231A033")
```

```python
positive_tweets=pd.DataFrame(all_positive_twitts,columns=["Tweets"]).head()
print("Shaikh Abdullah - 231A033")
```

```python
negative_tweets=pd.DataFrame(all_negative_twitts,columns=["Tweets"]).head()
print("Shaikh Abdullah - 231A033")
```

```python
positive_tweets["Sentiment"]='Positive'
negative_tweets["Sentiment"]='Negative'
positive_tweets
print("Shaikh Abdullah - 231A033")
```

```python
all_tweets=pd.concat([positive_tweets,negative_tweets],ignore_index=True)
all_tweets
print("Shaikh Abdullah - 231A033")
```

```python
all_tweets=all_tweets.sample(frac=1,random_state=42).reset_index(drop=True)
all_tweets
print("Shaikh Abdullah - 231A033")
```

```python
tweet1=all_tweets.iloc[5]
tweet1["Tweets"]
print("Shaikh Abdullah - 231A033")
```

```python
y_positiv=(all_tweets['Sentiment'] == 'Positive').sum()
print("Shaikh Abdullah - 231A033")
```

```python
y_negative=(all_tweets['Sentiment'] == 'Negative').sum()
print("Shaikh Abdullah - 231A033")
```

```python
import seaborn as sns

sentiment_counts = pd.DataFrame({'Sentiment': ['Positive', 'Negative'], 'Count': [y_positiv, y_negative]})

plt.figure(figsize=(7, 5))
sns.barplot(x='Sentiment', y='Count', hue='Sentiment', data=sentiment_counts, palette=['green', 'red'], legend=False)
plt.title('Distribution of Tweet Sentiments')
plt.xlabel('Sentiment')
plt.ylabel('Number of Tweets')
plt.show()
print("Shaikh Abdullah - 231A033")
```

```python
import re
import string

from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
from nltk.tokenize import TweetTokenizer
print("Shaikh Abdullah - 231A033")
```

```python
tweet2=re.sub(r'^RT[\S]+', '', tweet1["Tweets"])
tweet2=re.sub(r'https?:\/\/\S+', '', tweet2)
tweet2=re.sub(r'#', '', tweet2)
tweet2=re.sub(r'@\w+', '', tweet2)
tweet2
print("Shaikh Abdullah - 231A033")
```

```python
tokenizer =TweetTokenizer(preserve_case=False,strip_handles=True,reduce_len=True)
tweet_tokens=tokenizer.tokenize(tweet2)
print()
print('Tokenized string:\n')
print(tweet_tokens)
print("Shaikh Abdullah - 231A033")
```

```python
stowords_english=stopwords.words('english')
print('Stop words\n')
print(stowords_english)

print('\nPunctuation\n')
print(string.punctuation)
print("Shaikh Abdullah - 231A033")
```

```python
tweets_clean=[]
for words in tweet_tokens:
    if(words not in stowords_english and words not in string.punctuation):
        tweets_clean.append(words)
print(tweets_clean)
print("Shaikh Abdullah - 231A033")
```
